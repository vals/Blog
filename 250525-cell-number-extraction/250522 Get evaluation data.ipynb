```python
import pandas as pd
```

```python
data = pd.read_csv('https://nxn.se/single-cell-studies/data.tsv', sep = '\t', storage_options = {'User-Agent': ''})
```

```python
data.head().T
```

```python
df = (
    data
    .query('Measurement == "RNA-seq"')
    .dropna(subset=['Number of reported cell types or clusters'])
    .query('Journal != "bioRxiv"')
    .sample(100)
    .sort_values('Date')
)
```

```python
df.to_csv('250522.100papers.tsv', sep = '\t', index = False)
```
