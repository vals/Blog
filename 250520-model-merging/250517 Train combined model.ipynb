```python
import anndata
import plotnine as p
import scvi
```

```python
brain = anndata.io.read_h5ad('250508.brain_train.GSE212576.h5ad')
brain
```

```python
liver = anndata.io.read_h5ad('250508.liver_train.GSE166504.h5ad')
liver
```

```python
combined = anndata.concat([brain, liver])
```

```python
scvi.model.SCVI.setup_anndata(combined)
```

```python
combined_model = scvi.model.SCVI(
    combined,
    gene_likelihood = 'nb'
)
```

```python
combined_model.train(max_epochs = 10)
```

```python
combined_model.save(
    '250517.combined_model.10e.model',
    overwrite = True
)
```
