```python
import anndata
import plotnine as p
import scvi
```

```python
brain = anndata.io.read_h5ad('250508.brain_train.GSE212576.h5ad')
brain
```

```python
scvi.model.SCVI.setup_anndata(brain)
```

```python
brain_model = scvi.model.SCVI(
    brain,
    gene_likelihood = 'nb'
)
```

```python
brain_model.train(max_epochs = 20)
```

```python
brain_model.save(
    '250508.brain.model',
    overwrite = True
)
```
